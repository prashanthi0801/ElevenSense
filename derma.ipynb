```python
data=data.dropna(subset=["age_approx"])
# data["age_approx"].isnull().sum()
data["anatom_site_general"].fillna("unknown",inplace=True)
data["dermoscopic_type"].fillna("unknown",inplace=True)
data.drop(columns=["anatom_site_special"],inplace=True)
data["diagnosis_3"].fillna("unknown", inplace=True)
data["diagnosis_confirm_type"].fillna("unknown",inplace=True)
data["family_hx_mm"].fillna("unknown",inplace=True)
data["fitzpatrick_skin_type"].fillna(
    "unknown",
    inplace=True
)
data["personal_hx_mm"].fillna("unknown",inplace=True)
data["sex"].fillna("unknown",inplace=True)



```

```python
data.to_csv('cleaned_data1.csv', index=False)
# print("Cleaned data saved to cleaned_data.csv
```

```python
data.isnull().sum()
```

```python
data["concomitant_biopsy"] = data["concomitant_biopsy"].astype(int)
DROP_COLS = [
    "patient_id",
    "lesion_id",
    "isic_id",
    "attribution",
    "copyright_license",
    "anatom_site_general"

]

data.drop(columns=DROP_COLS , inplace=True)
data_clean=data
```

```python
y = data_clean["benign_malignant"].map({"benign": 0, "malignant": 1})
X = data_clean.drop(columns=["benign_malignant"])
```

```python
X.isnull().sum()
# x.info()
```

```python
from sklearn.feature_selection import mutual_info_classif
import pandas as pd
X_encoded = pd.get_dummies(X,drop_first=True)
mi_scores = mutual_info_classif(X_encoded,y,random_state=42)
mi_df = pd.DataFrame({
    "Feature": X_encoded.columns,
    "Information_Gain": mi_scores
}).sort_values(by="Information_Gain", ascending=False)
print(X_encoded.columns.tolist())
mi_df.head(10)
# mi_df.shape
```

```python
columns_with_unknown_encoding = [col for col in X_encoded.columns if 'unknown' in col]
print("One-hot encoded columns related to 'unknown' values:")
columns_with_unknown_encoding
```

```python
mi_df[mi_df['Feature'].isin(columns_with_unknown_encoding)]
```

```python
from sklearn.feature_selection import chi2

X_cat = pd.get_dummies(X.select_dtypes(include="object"))

chi_scores, p_values = chi2(X_cat, y)

chi_df = pd.DataFrame({
    "Feature": X_cat.columns,
    "Chi2_Score": chi_scores,
    "p_value": p_values
}).sort_values(by="Chi2_Score", ascending=False)

chi_df.head(10)
```

```python
top10_mi_features = set(
    mi_df
    .sort_values(by="Information_Gain", ascending=False)
    .head(10)["Feature"]
)
top10_mi_features
```

```python
top10_chi_features = set(
    chi_df[chi_df["p_value"] < 0.05]
    .sort_values(by="Chi2_Score", ascending=False)
    .head(10)["Feature"]
)
top10_chi_features
```

```python
chi_df[chi_df['Feature'].isin(columns_with_unknown_encoding)]
```

```python
common_features = list(top10_mi_features.intersection(top10_chi_features))
common_features
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.3, random_state=42)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression(max_iter=1000)
model.fit(X_train,y_train)
```

```python
from sklearn.metrics import classification_report, roc_auc_score

y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
```

```python

# LR for directional intuition
# +ve increases odds of malignancy
# -ve decreases odds of malignancy
```

```python
coef_df = pd.DataFrame({
    'feature': X_encoded.columns,
    'coefficient': model.coef_[0]
}).sort_values(by='coefficient', ascending=False)

coef_df
```

```python
from sklearn.ensemble import RandomForestClassifier
model=RandomForestClassifier(n_estimators=300,random_state=42, class_weight='balanced')
model.fit(X_train,y_train)
```

```python
model_importance = pd.DataFrame({
    'feature': X_encoded.columns,
    'importance': model.feature_importances_
}).sort_values(by='importance', ascending=False)

model_importance
```
